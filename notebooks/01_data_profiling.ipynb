```python
departure = pd.to_datetime(flights["departure_time"], errors="coerce")
arrival = pd.to_datetime(flights["arrival_time"], errors="coerce")
```

```python
print("Invalid departure times:", departure.isna().sum())
print("Invalid arrival times:", arrival.isna().sum())
```

```python
flights[arrival < departure][
    ["flight_id", "departure_time", "arrival_time", "duration"]
]
```

```python
calculated_duration = arrival - departure
```

```python
calculated_duration.head()
```

```python
comparison = flights[
    ["flight_id", "departure_time", "arrival_time", "duration"]
].copy()

comparison["calculated_duration"] = calculated_duration

comparison.head(10)
```
